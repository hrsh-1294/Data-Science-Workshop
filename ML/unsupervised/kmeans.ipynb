```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler

# Make sample data
X, y = make_moons(n_samples=300, noise=0.05, random_state=42)

# Scale the data (important for DBSCAN)
X_scaled = StandardScaler().fit_transform(X)

# Fit DBSCAN
db = DBSCAN(eps=0.3, min_samples=5)
db.fit(X_scaled)
labels = db.labels_  # cluster labels (-1 = noise)

# Plot DBSCAN result
plt.figure(figsize=(8, 5))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='viridis', s=50)

# Mark noise (label = -1) with black color
plt.scatter(
    X_scaled[labels == -1, 0],
    X_scaled[labels == -1, 1],
    color='black', s=50, label='Noise'
)

plt.title('DBSCAN Clustering')
plt.xlabel('Feature 1 (scaled)')
plt.ylabel('Feature 2 (scaled)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import scipy.cluster.hierarchy as sch

# --- 1. Generate some sample data ---
# Using make_blobs to create clearly separated groups
X, y = make_blobs(n_samples=25, centers=3, n_features=2, random_state=42)

# --- 2. Perform Hierarchical Clustering and Plot the Dendrogram ---
plt.figure(figsize=(12, 8))
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Data Point Index')
plt.ylabel('Distance')

# The linkage function calculates the distances and performs the clustering
# The dendrogram function takes this result and plots it
linked = sch.linkage(X, method='ward')
dendrogram = sch.dendrogram(linked,
                            orientation='top',
                            labels=np.arange(1, len(X) + 1), # Custom labels for points
                            distance_sort='descending',
                            show_leaf_counts=True)

# Add a horizontal line to suggest a cut
plt.axhline(y=15, color='r', linestyle='--')
plt.annotate('Cut here for 3 clusters', xy=(100, 16))

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
