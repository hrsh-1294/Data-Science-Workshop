```python
import os

# Create a folder named "new_folder"
os.mkdir("operating")
print("Directory 'operating' created successfully!")
```

```python
cwd = os.getcwd()
print("Current working directory:", cwd)
```

```python
items = os.listdir(".")
print("Items in current directory:", items)
```

```python
os.rmdir("operating")
print("Directory removed!")
```

```python
if os.path.exists("operating"):
    print("Folder exists!")
else:
    print("Folder not found.")
```

```python
path = "example.txt"

if os.path.isfile(path):
    print(f"{path} is a file.")
elif os.path.isdir(path):
    print(f"{path} is a directory.")
else:
    print("Path not found.")
```

```python
# Get a specific environment variable
print(os.environ.get("USERNAME", "Unknown User"))

# Print all environment variables
for key, value in os.environ.items():
    print(f"{key}: {value}")
```

```python
# Join parts into a valid path
path = os.path.join("parent", "child", "file.txt")
print("Joined path:", path)

# Split path into directory and file
dir_name, file_name = os.path.split(path)
print("Directory:", dir_name)
print("File:", file_name)
```

```python
abs_path = os.path.abspath("file.txt")
print("Absolute path:", abs_path)
```

```python
import os

file_name = "file.txt"  # change this to your file name

if os.path.exists(file_name):
    os.remove(file_name)
    print(f"✅ File '{file_name}' deleted successfully!")
else:
    print("❌ File not found!")
```
