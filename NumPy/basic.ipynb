```python
arr = np.arange(60)
print(arr.size)
print(arr.ndim)
print(arr.shape)

#convert to 2d
arr2d = arr.reshape(12,5) #rows, columns
print(arr2d)
print(arr2d[0][4]) #indexing
```

```python
print(arr2d[0:2, 0:4]) #slicing
```

```python
arr2d[2][3] = 20
arr2d
```

```python
# arr2d = arr2d.reshape(5,12) #transpose
#or

arr2d = arr2d.T
print(arr2d)
print(arr2d.shape)
print(arr2d.ndim)
print(arr2d.dtype)
```

```python
arr3d = arr2d.reshape(3,4,5)
print(arr3d)

#3d to 1d
arr1d = arr3d.flatten() #or arr3d.ravel() -> different from flatten as it returns a view and flatten returns a copy
print(arr1d)
```
